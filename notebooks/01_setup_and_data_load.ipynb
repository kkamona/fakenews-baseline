```python

import os
import pandas as pd

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
DATA_RAW_DIR = os.path.join(PROJECT_ROOT, "data", "raw")
DATA_PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data", "processed")

os.makedirs(DATA_RAW_DIR, exist_ok=True)
os.makedirs(DATA_PROCESSED_DIR, exist_ok=True)

print("PROJECT_ROOT:", PROJECT_ROOT)
print("DATA_RAW_DIR:", DATA_RAW_DIR)

RAW_CSV_PATH = os.path.join(DATA_RAW_DIR, "WELFake_Dataset.csv")

if not os.path.exists(RAW_CSV_PATH):
    raise FileNotFoundError(
        f"Raw CSV not found at: {RAW_CSV_PATH}\n"
        "Download WELFake and place the CSV in data/raw/ with the name WELFake_Dataset.csv"
    )

df = pd.read_csv(RAW_CSV_PATH)

print("\n Loaded raw dataset.")
print("Shape:", df.shape)
print("Columns:", list(df.columns))

print("\nSample rows:")
display(df.head(3))

print("\nMissing values:")
display(df.isna().sum().sort_values(ascending=False).head(10))
```
